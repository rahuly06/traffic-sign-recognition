Data pre-processing for the YOLO

```python
import os
import cv2
```

```python
# Paths
input_dir = r"C:\Users\rahul\OneDrive\Desktop\Rahul\Study\Projects\traffic-sign-recognition\data\archive_YOLO\TestIJCNN2013\TestIJCNN2013Download"      # folder with .ppm images
output_dir = r"C:\Users\rahul\OneDrive\Desktop\Rahul\Study\Projects\traffic-sign-recognition\data\archive_YOLO\TestIJCNN2013\TestIJCNN2013Download\Test_data\Images" # folder to save .jpg images
os.makedirs(output_dir, exist_ok=True)
```

```python
# Convert each .ppm to .jpg
for filename in os.listdir(input_dir):
    if filename.endswith(".ppm"):
        ppm_path = os.path.join(input_dir, filename)
        img = cv2.imread(ppm_path)
        jpg_filename = os.path.splitext(filename)[0] + ".jpg"
        jpg_path = os.path.join(output_dir, jpg_filename)
        cv2.imwrite(jpg_path, img)

print("Conversion completed!")
```

```python
# path for text files
# Paths
images_dir = r"C:\Users\rahul\OneDrive\Desktop\Rahul\Study\Projects\traffic-sign-recognition\data\archive_YOLO\TestIJCNN2013\TestIJCNN2013Download\Test_data\Images"
gt_file = r"C:\Users\rahul\OneDrive\Desktop\Rahul\Study\Projects\traffic-sign-recognition\data\archive_YOLO\TrainIJCNN2013\TrainIJCNN2013\gt.txt"
labels_dir = r"C:\Users\rahul\OneDrive\Desktop\Rahul\Study\Projects\traffic-sign-recognition\data\archive_YOLO\TrainIJCNN2013\TrainIJCNN2013\Train_data\TXT"
os.makedirs(labels_dir, exist_ok=True)
```

```python
# read gt.txt
# Read ground truth file
with open(gt_file, "r") as f:
    lines = f.readlines()
```

```python
# Group annotations per image
annotations = {}
for line in lines:
    parts = line.strip().split(";")
    img_name = parts[0].replace(".ppm", ".jpg")  # match converted jpg filenames
    left, top, right, bottom = map(int, parts[1:5])
    class_id = int(parts[5])

    if img_name not in annotations:
        annotations[img_name] = []
    annotations[img_name].append((left, top, right, bottom, class_id))
```

```python
# Convert each image's annotations
for img_name, objs in annotations.items():
    img_path = os.path.join(images_dir, img_name)
    if not os.path.exists(img_path):
        print(f"⚠️ Skipping {img_name}, not found in Images folder")
        continue

    # Get image size (needed for normalization)
    img = cv2.imread(img_path)
    h, w, _ = img.shape

    label_path = os.path.join(labels_dir, img_name.replace(".jpg", ".txt"))

    with open(label_path, "w") as f:
        for left, top, right, bottom, class_id in objs:
            # Convert to YOLO format
            x_center = ((left + right) / 2) / w
            y_center = ((top + bottom) / 2) / h
            width = (right - left) / w
            height = (bottom - top) / h

            f.write(f"{class_id} {x_center:.6f} {y_center:.6f} {width:.6f} {height:.6f}\n")

print("✅ YOLO label conversion completed!")
```
